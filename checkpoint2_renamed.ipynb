## Data Exploration: Kruskal-Wallis test

Motivation for Selection: We must first consider the fact that this is a dataset made up of states varying in population
size. Naturally, this influences the C02 emissions value, and the likelihood of outliers. For example, California's
total CO2 for all fuels is roughly 304.47 million metric tons while Alaska's is 12.78 in the year 1974. ANOVA compares group means, but reacts
poorly to the presence of outliers, so we will need a test that can handle this issue. With a sample size of 616, the lack of a normal distributed
dataset and having categorical variables (e.g., sector name) makes the selection of a non-parametric test appealing. Kruskal-Wallis
achieves our goal of comparing categorical variables by using group medians, which is an appropriate choice considering
our circumstances. First, we can create separate dataframes for each sector (in the year 2021) from the dictionary, which altogether
make up a mid-sized dataset and gives a narrower analysis of the original dataset. We want to be carefully of the rows that contain
aggregated values (i.e., total from all sectors) to ensure the validity of our results.

```python

import matplotlib.pyplot as plt
from scipy import stats

modified_df = df[(df['sector-name'] != 'Total from all sectors') & (df['fuel-name'] != 'All Fuels') &(df['year'] == 2021)]

print('Samples:',modified_df.shape[0])


sector_dict = {sector_type: group for sector_type, group in modified_df.drop_duplicates().groupby('sector-name')}


resid= sector_dict['Residential']
comm= sector_dict['Commercial']
transport= sector_dict['Transportation']
elec = sector_dict['Electric Power']
industrial = sector_dict['Industrial']

display('Residential Data:', resid.head())
display('Commercial Data: ', comm.head())
display('Transportation Data: ',transport.head())
display('Electrical Data: ',elec.head())
display('Industrial Data: ',industrial.head())
```

Hypothesis Testing & Plotting Data: Now, we must create our null and alternative hypotheses. This is based
on whether the medians of the sectors will be the same or different, specifically if CO2 values will differentiate depending on the
sector. Next, the data is presented in the form of a boxplot with the y-axis scaled by log to account for the fact that
outliers can cause the plot to appear compressed. The ranks of each sector are analyzed below.

HO: There is no difference in medians of the sectors for CO2 emissions.

HA:There is at least one differing median among the sectors for CO2 emissions.

```python
stat, p_value = stats.kruskal(resid['value'],comm['value'],transport['value'],elec['value'], industrial['value'])
plt.figure(figsize=(10,5))
plt.boxplot([resid['value'],comm['value'],transport['value'],elec['value'], industrial['value']], labels=['Residential','Commercial', 'Transportation','Electrical','Industrial'])
plt.yscale('log') #to increase readibility as many values lie close to 0 
plt.title(f'Comparing Sector Ranks Using CO2 Emissions')
plt.ylabel(f'CO2 Value')
plt.show()
```

P-value analysis:

Assume the alpha value is 0.05.

We see that our p-value is signficantly smaller than our alpha value, so we are able to reject

the null hypothesis and discern that at least one group's median differs from the others. Most importantly,
the long distance from alpha highlights that the statistical significance has little ambiguity.

```python
print(f'The p-value is {p_value}')
```

Conclusion & Plot-Analysis:
We notice that the transportation's and electrical's box lengths stand out amongst the others. The longer sizing tells us
that there is greater variability in CO2 values for transportation and electrical, meaning that there is a bigger buffer
space for outliers. In contrast, the other sectors showcase higher consistency and thus, are more sensitive/quick to label
data points as outliers. Additionally, we notice that transportation, electrical, and industrial have outliers placed near
the very top of the plot. These findings can be informative for policymakers, as the scale and factors contributing
to the variability of electrical and transportation ranks need more researching, since there isn't a level of uniformity
that implies that there are dominant factors as to why CO2 is high.

## Chi Square Test of Independence

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

df = pd.read_csv('emissions.csv')
df['year'] = df['year'].astype(int)
df['value'] = df['value'].astype(float) #CO2 values are floats
df['sector-name'] = df['sector-name'].str.replace(' carbon dioxide emissions', '', regex=False)

clean_df = df[(df['sector-name'] != 'Total from all sectors') & (df['fuel-name'] != 'All Fuels') & (df['year'] == 2021)].copy()
#above code makes sure it keeps only year 2021 so it filters out Total from all sectors and All Fuels 

median = clean_df['value'].median() #chi-square needs categorical data 
clean_df['emission_level'] = np.where(clean_df['value'] > median, 'High Emitter', 'Low Emitter')
#if a row's CO2 value is greater than the median, it labels it 'High Emitter' else 'Low Emitter'

table = pd.crosstab(clean_df['sector-name'], clean_df['emission_level'])
print(table)

test_statistic, p, dof, expected = stats.chi2_contingency(table) 
print(f"\nChi-square statistic: {test_statistic:.2f}")
print(f"Degrees of freedom: {dof}")
print(f"p-value: {p:.4e}")
divide_counts_by_row_total = table.div(table.sum(axis=1), axis=0)

divide_counts_by_row_total.plot(kind='bar', stacked=True, figsize=(10, 6), color=['#d62728', '#2ca02c'])
plt.title('Proportion of High vs. Low Emitters Across Sectors (2021)')
plt.xlabel('Sector')
plt.ylabel('Proportion of States')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Emission Level', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

Hypothesis Testing and Conclusion:

Null Hypothesis: The proportion of High vs. Low emitters is independent of the sector.
Alternative Hypothesis: The proportion of High vs. Low emitters depends on the sector.

We used the Chi-Square Test of Independence on the 2021 data, using the median CO2 value to categorize emissions into 
"High" and "Low". We got a test statistic of 31.98 with 4 degrees of freedom and a p-value of 1.93e-06. 

Since the p-value is significantly lower than the alpha of 0.05, we reject the null hypothesis. There is a statistically significant 
association between the sector and the likelihood of a state having high CO2 emissions. 

As you can see in the stacked bar chart, sectors like Industrial (91 high vs 54 low) and Electric Power (77 high vs 67 low) 
are skewed heavily towards being high emitters. Commercial and Residential are predominantly low emitters. 
This shows that Industrial and Electric Power are the primary contributors to extreme CO2 emissions.  

## Chow Test

### Introductory data exploration

I only realized this was not part of the course until after it was mostly complete, so I decided to include it anwyays.*

To begin, let's take a look at the total emissions by year:

```python
import matplotlib.pyplot as plt

totals = df[(df["sector-name"] == "Total from all sectors") & (df["fuel-name"] == "All Fuels")].copy()
annual_emissions = (totals.groupby("year")["value"].sum().reset_index().rename(columns={"value": "total_co2"}))
#print(annual_emissions.head())

# plot annual emissions data
plt.figure(figsize=(12,6))

x = annual_emissions['year']
y = annual_emissions['total_co2']

# this plot is not that glorious admittedly
plt.scatter(x, y)

plt.title('Annual CO2 Emissions')
plt.xlabel('Year')
plt.ylabel('Total CO2 Emissions (Million Metric Tons)')

plt.plot()

```

We observe that there is a peak at 2005, and that from 1982 to 2005 there seems to be a steady incline and from 2005 onwards a steady decline. We want to know if this change in slope (at 2005) is statistically significant, and if so by how much the slope shifts, so we perform a Chow Test for structural break:

HO: The change in slope at 2005 is not statistically significant 

HA: The change in slope at 2005 is statistically significant

We will only include years after 1981, since that is where the main upwards slope begins.

### Hypothesis testing and graphing

```python
import seaborn as sns
import statsmodels.formula.api as smf

# years after 1981
annual_emissions = annual_emissions[annual_emissions["year"] > 1981].reset_index(drop=True)

# the chow test uses a dummy variable (is_post_2005), and honestly I'm still not entirely sure why, but it works lol
annual_emissions["year_centered"] = annual_emissions["year"] - 2005
annual_emissions["is_post_2005"] = (annual_emissions["year"] > 2005).astype(int) 

# fit interaction model
chow_model = smf.ols("total_co2 ~ year_centered * is_post_2005", data=annual_emissions).fit()

# get params
slope_pre_2005 = chow_model.params["year_centered"]
slope_shift = chow_model.params["year_centered:is_post_2005"]
slope_post_2005 = slope_pre_2005 + slope_shift

p_val = chow_model.pvalues["year_centered:is_post_2005"]

print("Results:")
print(f"1982–2005 Slope: {slope_pre_2005:+.2f} MMT/year")
print(f"2006–2021 Slope: {slope_post_2005:+.2f} MMT/year")
print(f"Slope Shift: {slope_shift:+.2f} MMT/year")
print(f"p-value: {p_val:.4e}")

# glorious plotting (using seaborn to help with the linear regression viz)
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(annual_emissions["year"], annual_emissions["total_co2"], color="#2b5c8f", s=55, zorder=3, label="Observed CO2 Emissions")

upwards = annual_emissions[annual_emissions["year"] <= 2005]
downwards = annual_emissions[annual_emissions["year"] >= 2005]

# linear regression for upwards/downwards slope sections
sns.regplot(data=upwards, x="year", y="total_co2", ax=ax, color="#1f77b4", scatter=False, line_kws={"linewidth": 2.5, "label": f"1983–2005 Trend ({slope_pre_2005:+.1f} MMT/yr)"})
sns.regplot(data=downwards, x="year", y="total_co2", ax=ax, color="#d95f02", scatter=False, line_kws={"linewidth": 2.5, "label": f"2006–2021 Trend ({slope_post_2005:+.1f} MMT/yr)"})

# structural break line
ax.axvline(2005, color="gray", linestyle="--", linewidth=1.5, alpha=0.7, label="2005 Structural Break")

# textbox
stats_text = (f"Slope Shift: {slope_shift:+.2f} MMT/yr\n" f"p-value: {p_val:.4e}\n")
ax.text(0.02, 0.97, stats_text, transform=ax.transAxes, fontsize=11, verticalalignment="top", bbox=dict(boxstyle="round,pad=0.5", facecolor="white", edgecolor="#cccccc"))

# other text
ax.set_title("U.S. Yearly CO2 Emissions: Chow Test for Structural Break", fontsize=13, fontweight="bold")
ax.set_xlabel("Year", fontsize=10)
ax.set_ylabel("Total CO2 Emissions (MMT)", fontsize=11)
ax.legend(loc="lower left")

plt.tight_layout()
plt.show()
```

### Conclusions



We see from the results of the test the p-value is increadibly small (~2.5e-19), which means that the "structural break," or the change in slope at 2005 observable in the original graph and represented by the two linear regression models (blue and orange lines), is indeed statistically significant. With this in mind, we successfully reject HO.

The Chow Test also tells us how much the slope has shifted; The results show a downward shift of ~277 MMT/year. Assuming we knew that emissions policies were enacted in 2005, we could use this to tell whether they were successfuly or not.

## T-Test

Data Exploration: Paired T-Test

Motivation for Selection: due to states drastically differing sizes which creates outliers in the data, a paired t-test could control this by comparing each states coal and natural gas emissions against each other in the same year. We used the electric power sector in 2020, since coal and natural gas are the main competing fuels in that sector. States with no reported value for a fuel were treated as zero emissions.

```python
electric_df = df[(df['sector-name'] == 'Electric Power') & 
            (df['fuel-name'].isin({'Coal','Natural Gas'})) & 
            (df['year'] == 2020) & 
            (df['state-name'] != 'United States')]
            #dropping the united states data to avoid double counting

pivot = electric_df.pivot(index='state-name', columns='fuel-name', values='value')

#we're filling all null values to 0 as we checked which states that have null values
#it isnt that they have missing data, they just generate nearly no emissions for that year/state
pivot = pivot.fillna(0)
pivot.head(5)
```

Hypothesis Testing & Plotting Data: The null hypothesis is that, states coal and natural gas electric-power emissions in 2020 are the same. The alternative is that they differ. The data is shown as a scatter plot, each dot being a states coal emissions against its natural gas emissions. A dashed line shows where the two are equal. 

HO = The mean difference between natural gas and coal for states electric-power emissions in 2020 are the same

HA = The mean difference between natural gas and coal for states electric-power emissions in 2020 are different

```python
stat, p_value = stats.ttest_rel(pivot['Natural Gas'], pivot['Coal'])

plt.scatter(pivot['Coal'], pivot['Natural Gas'])
lim = pivot.max().max() * 1.05
plt.plot([0, lim], [0, lim], 'k--', label='gas = coal')
plt.xlabel('Coal emissions')
plt.ylabel('Natural gas emissions')
plt.title('Electric power CO2 by state, 2020')
plt.show()

print(f"p value is: {p_value}")
```


Assume the alpha value is 0.05.


Conclusion & Plot-Analysis:
We see that our p-value is above the alpha value so we fail to reject the null hypothesis. Meaning, there is no significant difference between states coal and natural gas emissions in 2020. We can see from the graph that on average coal emissions were higher, as there is more of a spread in points below the line compared to the tighter bunching above the line. The fuels this year roughly balance each other out, however 2020 was the first year of the pandemic which does explain the reduced demand for electricity. Below is a pre-pandemic comparison with 2019 using the same t-test method which shows higher coal emissions, and p value that would fail to reject the null hypothesis.

```python
electric_df = df[(df['sector-name'] == 'Electric Power') & 
            (df['fuel-name'].isin({'Coal','Natural Gas'})) & 
            (df['year'] == 2019) & 
            (df['state-name'] != 'United States')]

pivot = electric_df.pivot(index='state-name', columns='fuel-name', values='value')
pivot = pivot.fillna(0)

stat, p_value = stats.ttest_rel(pivot['Natural Gas'], pivot['Coal'])

plt.scatter(pivot['Coal'], pivot['Natural Gas'])
lim = pivot.max().max() * 1.05
plt.plot([0, lim], [0, lim], 'k--', label='gas = coal')
plt.xlabel('Coal emissions')
plt.ylabel('Natural gas emissions')
plt.title('Electric power CO2 by state, 2020')
plt.show()

print(f"p value is: {p_value}")
```
